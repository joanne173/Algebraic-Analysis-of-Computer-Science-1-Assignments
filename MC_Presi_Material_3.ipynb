```python
#Shrink to fit applied to vector
"""""
#include <vector>
#include <string>
using namespace std;
int main()
{
    vector<string> vec={"-", "-", "-", "-", "-", "-", "-", "*"};

    for (int i=0; i<=14; i++){
        //create the mechanism to add 2 stars and remove the beginning item

        for (string element : vec){
            cout<<element;
        }
        cout << vec.size()<<" "<< vec.capacity()<< endl;
        if (i<7){
            for (int j=1; j<=2; j++){
                vec.push_back("*");
            }
            for (int k=0; k<1; k++){
                vec.erase(vec.begin());
            }
        }
        else{
            for(int j=1; j<=2;j++){
                vec.pop_back();
            }
            for(int k=0; k<1; k++){
                vec.insert(vec.begin(),"-");
            }
        }
    vec.shrink_to_fit();
    }
}
//notice how in the output the capacity of the vector is dynamically changing with the size of said vector taking less memory space

-------*8 8
------***9 9
-----*****10 10
----*******11 11
---*********12 12
--***********13 13
-*************14 14
***************15 15
-*************14 14
--***********13 13
---*********12 12
----*******11 11
-----*****10 10
------***9 9
-------*8 8
"""""

```
